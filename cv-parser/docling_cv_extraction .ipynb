#  Extraction de CVs avec Docling
Ce notebook couvre :
- Installation & imports
- Configuration du dossier de CVs
- Extraction batch sur tous les CVs
- Tous les formats d'export (Markdown, JSON, texte brut, HTML)
- Extraction avancée : email, téléphone, liens, structures répétitives

---
##  Cellule 1 — Installation & Imports

```python
# Installation
# À exécuter une seule fois
%pip install -q docling pandas openpyxl

# Imports
from docling.document_converter import DocumentConverter
from pathlib import Path
import json
import re
import pandas as pd
from IPython.display import display, Markdown, HTML

print(" Tout est installé et importé.")
```

---
##  Cellule 2 — Configuration

```python
# Dossiers
DOSSIER_CVS   = Path("cv_exemple")      # PDF/DOCX
DOSSIER_OUT   = Path("cv_output")       # dossier de sortie auto-créé

DOSSIER_CVS.mkdir(exist_ok=True)
DOSSIER_OUT.mkdir(exist_ok=True)

# Convertisseur Docling
converter = DocumentConverter()

# Formats acceptés
EXTENSIONS = [".pdf", ".doc", ".docx"]

# Vérification : liste les CVs trouvés
cvs = [f for f in DOSSIER_CVS.iterdir() if f.suffix.lower() in EXTENSIONS]
print(f" Dossier   : {DOSSIER_CVS.resolve()}")
print(f" CVs trouvés : {len(cvs)}")
for cv in cvs:
    print(f"   └─ {cv.name}")
```

---
##  Cellule 3 — Extraction batch sur tous les CVs

```python
# Dictionnaire principal : { nom_fichier: result_docling }
resultats = {}
erreurs   = []

for cv_path in cvs:
    try:
        result = converter.convert(cv_path)
        resultats[cv_path.name] = result
        print(f" {cv_path.name}")
    except Exception as e:
        erreurs.append(cv_path.name)
        print(f" {cv_path.name} — Erreur : {e}")

print(f"\n {len(resultats)} CV(s) extraits avec succès, {len(erreurs)} erreur(s).")
```

---
##  Cellule 4 — Export Markdown

```python
#  Affichage du premier CV en Markdown
if resultats:
    premier = list(resultats.keys())[0]
    md_text = resultats[premier].document.export_to_markdown()

    print(f"── Markdown brut : {premier} ──")
    print(md_text[:1500], "...")

    # Rendu visuel dans le notebook
    display(Markdown("---\n**Rendu Markdown :**\n\n" + md_text[:1500]))

# Sauvegarde Markdown de tous les CVs
md_dir = DOSSIER_OUT / "markdown"
md_dir.mkdir(exist_ok=True)

for nom, result in resultats.items():
    out = md_dir / (Path(nom).stem + ".md")
    out.write_text(result.document.export_to_markdown(), encoding="utf-8")

print(f"\n Fichiers .md sauvegardés dans : {md_dir}")
```

---
##  Cellule 5 — Export Texte brut (.txt)

```python
# Affichage du premier CV en texte brut
if resultats:
    premier = list(resultats.keys())[0]
    txt = resultats[premier].document.export_to_text()
    print(f"── Texte brut : {premier} ──")
    print(txt[:1500], "...")

# Sauvegarde .txt de tous les CVs
txt_dir = DOSSIER_OUT / "txt"
txt_dir.mkdir(exist_ok=True)

for nom, result in resultats.items():
    out = txt_dir / (Path(nom).stem + ".txt")
    out.write_text(result.document.export_to_text(), encoding="utf-8")

print(f"\n Fichiers .txt sauvegardés dans : {txt_dir}")
```

---
##  Cellule 6 — Export JSON structuré

```python
#  Affichage de la structure JSON du premier CV
if resultats:
    premier = list(resultats.keys())[0]
    data    = resultats[premier].document.export_to_dict()
    print(f"── JSON structuré : {premier} ──")
    print(json.dumps(data, indent=2, ensure_ascii=False)[:2000], "...")

# Sauvegarde JSON de tous les CVs
json_dir = DOSSIER_OUT / "json"
json_dir.mkdir(exist_ok=True)

for nom, result in resultats.items():
    out = json_dir / (Path(nom).stem + ".json")
    with open(out, "w", encoding="utf-8") as f:
        json.dump(result.document.export_to_dict(), f, ensure_ascii=False, indent=2)

print(f"\n Fichiers .json sauvegardés dans : {json_dir}")
```

---
##  Cellule 7 — Export HTML

```python
# Affichage HTML du premier CV
if resultats:
    premier  = list(resultats.keys())[0]
    html_str = resultats[premier].document.export_to_html()
    print(f"── HTML : {premier} (aperçu) ──")
    print(html_str[:800], "...")

    # Rendu dans le notebook
    display(HTML(html_str))

# Sauvegarde HTML de tous les CVs
html_dir = DOSSIER_OUT / "html"
html_dir.mkdir(exist_ok=True)

for nom, result in resultats.items():
    out = html_dir / (Path(nom).stem + ".html")
    out.write_text(result.document.export_to_html(), encoding="utf-8")

print(f"\n Fichiers .html sauvegardés dans : {html_dir}")
```

---
##  Cellule 8 — Extraction avancée : Email, Téléphone, LinkedIn, GitHub

```python
# Patterns regex pour structures répétitives
PATTERNS = {
    "email":    r'[\w.+\-]+@[\w\-]+\.[\w.]+',
    "telephone": r'(?:\+33|0)[\s.\-]?[1-9](?:[\s.\-]?\d{2}){4}',  # format FR
    "linkedin": r'linkedin\.com/in/[\w\-]+',
    "github":   r'github\.com/[\w\-]+',
    "site_web": r'https?://[\w.\-/]+',
}

def extraire_champs(texte: str) -> dict:
    """Extrait les champs structurés d'un texte brut."""
    return {
        champ: list(set(re.findall(pattern, texte, re.IGNORECASE)))
        for champ, pattern in PATTERNS.items()
    }

# Application sur tous les CVs
champs_extraits = []

for nom, result in resultats.items():
    texte  = result.document.export_to_text()
    champs = extraire_champs(texte)
    champs["fichier"] = nom
    champs_extraits.append(champs)
    print(f"\n📄 {nom}")
    for k, v in champs.items():
        if k != "fichier" and v:
            print(f"   {k:12} → {v}")

df_champs = pd.DataFrame(champs_extraits)
display(df_champs)
```

---
##  Cellule 9 — Extraction avancée : Sections répétitives (Expérience, Formation, Compétences)

```python
# Mots-clés de sections courants dans les CVs (FR + EN)
SECTIONS_KEYWORDS = {
    "experience":   ["expérience", "experience", "emploi", "poste", "parcours professionnel"],
    "formation":    ["formation", "éducation", "education", "diplôme", "diplome", "études", "etudes"],
    "competences":  ["compétences", "competences", "skills", "technologies", "outils", "langages"],
    "langues":      ["langues", "languages"],
    "projets":      ["projets", "projects", "réalisations"],
    "certifications": ["certifications", "certificats", "certifications"],
}

def detecter_sections(texte: str) -> dict:
    """Détecte quelles sections sont présentes dans le CV."""
    texte_lower = texte.lower()
    return {
        section: any(kw in texte_lower for kw in kws)
        for section, kws in SECTIONS_KEYWORDS.items()
    }

def extraire_section_brute(texte: str, keywords: list) -> str:
    """Extrait le texte brut d'une section en cherchant ses keywords."""
    lignes  = texte.split("\n")
    capture = False
    buffer  = []

    for ligne in lignes:
        ll = ligne.lower().strip()
        # Début de section
        if any(kw in ll for kw in keywords) and len(ll) < 60:
            capture = True
            buffer  = [ligne]
            continue
        # Fin de section : nouvelle section détectée
        if capture:
            toutes_kws = [kw for kws in SECTIONS_KEYWORDS.values() for kw in kws]
            if any(kw in ll for kw in toutes_kws) and len(ll) < 60 and buffer:
                break
            buffer.append(ligne)

    return "\n".join(buffer).strip()

# Affichage des sections détectées pour chaque CV
recap_sections = []

for nom, result in resultats.items():
    texte   = result.document.export_to_text()
    present = detecter_sections(texte)
    row     = {"fichier": nom}
    row.update({k: "" if v else "—" for k, v in present.items()})
    recap_sections.append(row)
    print(f"\n {nom}")
    for section, found in present.items():
        if found:
            extrait = extraire_section_brute(texte, SECTIONS_KEYWORDS[section])
            print(f"\n  [{section.upper()}]")
            print(extrait[:400] + ("..." if len(extrait) > 400 else ""))

display(pd.DataFrame(recap_sections))
```

---
## Cellule 10 — Export final consolidé (CSV + Excel)

```python
# Tableau final : champs extraits + sections détectées
df_contact  = pd.DataFrame(champs_extraits)
df_sections = pd.DataFrame(recap_sections)

df_final = df_contact.merge(df_sections, on="fichier", how="outer")

# Ajouter le texte brut complet
df_final["texte_brut"] = df_final["fichier"].apply(
    lambda nom: resultats[nom].document.export_to_text() if nom in resultats else ""
)
df_final["markdown"] = df_final["fichier"].apply(
    lambda nom: resultats[nom].document.export_to_markdown() if nom in resultats else ""
)

# Export CSV
csv_path = DOSSIER_OUT / "cvs_extraits.csv"
df_final.to_csv(csv_path, index=False, encoding="utf-8")
print(f" CSV    → {csv_path}")

# Export Excel
excel_path = DOSSIER_OUT / "cvs_extraits.xlsx"
df_final.drop(columns=["texte_brut", "markdown"]).to_excel(excel_path, index=False)
print(f" Excel  → {excel_path}")

print("\n── Aperçu du tableau final ──")
display(df_final.drop(columns=["texte_brut", "markdown"]))
```
